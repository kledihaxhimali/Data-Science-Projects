# DATA COLLECTION

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
credits =pd.read_csv('tmdb_5000_credits.csv')
movies= pd.read_csv('tmdb_5000_movies.csv')
```

```python
credits.head(n=7)
```

```python
movies.head(5) # movie and credits have the same id so we can merge them
```

# DATA PREPROCESSING




```python
credits_updated = credits.rename(columns={'movie_id':'id'})
credits_updated
```

```python
movies.shape
```

```python
credits_updated.shape
```

```python
movies_merge=movies.merge(credits_updated, on='id')
movies_merge.shape
```

```python
movies_cleaned= movies_merge.drop(columns=['title_y', 'homepage', 'title_x', 'status', 'production_countries'])
movies_cleaned.head(2)
```

# DATA ANALYSIS + FEATURE ENGINEERING

```python
V= movies_cleaned['vote_count']
R= movies_cleaned['vote_average']
C= movies_cleaned['vote_average'].mean()
m= movies_cleaned['vote_count'].quantile(0.70)
```

```python
C
```

```python
movies_cleaned[['original_title', 'vote_count']]
```

```python
sns.boxplot(data=movies_cleaned, x='vote_count')
m=1900
```

```python
m =movies_cleaned['vote_count'].quantile(q= 0.9)#nearest threshold to 1900
```

```python
filter= movies_cleaned[movies_cleaned['vote_count']>=m]
filter
```

```python
v= filter['vote_count']
R= filter['vote_average']
C= filter['vote_average'].mean()
```

```python
filter['weighted_average']=(R*v+C*m)/(v+m)
sorted_ranking=filter.sort_values('weighted_average', ascending=False).head(10)
sorted_ranking[['original_title', 'vote_average', 'vote_count', 'weighted_average', "popularity"]]
```

```python
sns.barplot(x=sorted_ranking['weighted_average'], y=sorted_ranking['original_title'])
```

# MODEL BUILDING

```python
popularity= filter.sort_values('popularity', ascending=False)
popularity.set_index('original_title')['popularity'][0:10].plot.barh()
```

# DATA NORMALIZATION

```python
from sklearn.preprocessing import MinMaxScaler
scaling= MinMaxScaler()
scaling_values=scaling.fit_transform(filter[['weighted_average','popularity']])
filter[['weighted_average_scaled', 'popularity_scaled']]=scaling_values
filter.head(3)
```

```python
filter['score_mix']= filter['weighted_average_scaled']*0.5+filter['popularity_scaled']*0.5
rank=filter.sort_values('score_mix', ascending=False).head(10)
```

```python
import plotly.express as px
fig=px.bar(data_frame=rank, x='original_title', y='score_mix')
fig.show()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfv= TfidfVectorizer(min_df=3, max_features=None, ngram_range=(1,3), stop_words='english')
tfv_matrix= tfv.fit_transform(data['overview'])
tfv_matrix.shape
```

SIMILARITY SCORES

```python
data['overview']
```

```python
from sklearn.metrics.pairwise import sigmoid_kernel
sig= sigmoid_kernel(tfv_matrix, tfv_matrix)
sig
```

```python
pd.DataFrame(sig)
```

```python
indicies= pd.Series(data = data.index, index=data['original_title'])
indicies['John Carter']
```

```python
sig[4]
```

```python
list(enumerate(sig[indecies['John Carter']]))
sigma= sorted(list(enumerate(sig[indecies['John Carter']])), key=lambda x: x[1], reverse=True)
sigma [0:10]
```

```python
ind= [i[0] for i in sigma[0:10]]
ind
dataframe= data.reset_index()
dataframe['original_title'][ind]
```

MODEL DEPLOYMENT

```python
def give_recommendations(movie_title, model):
  indicies= pd.Series(data = data.index, index=data['original_title'])
  idx=indicies[movie_title]
  model_score= list(enumerate(model[idx]))
  model_scores_sorted= sorted(model_score, key=lambda x: x[1], reverse=True)
  model_scores_10= model_scores_sorted[1:11]
  movie_indices_10= [i[0] for i in model_scores_10]
  return dataframe['original_title'][movie_indices_10]

give_recommendations("Avatar", sig)
```
